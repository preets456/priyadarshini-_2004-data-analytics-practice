```python
df['customer_age'].mean()
```

```python
df.groupby('gender')['customer_age'].mean().plot(kind='bar',color='b')
plt.title('AVG OF UNEMPLOYED PEOPLE')
plt.xlabel('GENDER')
plt.ylabel('AGE')
plt.show()
```

```python
plt.scatter(df['customer_age'],df['region'],color='maroon',marker='.')
plt.xlabel('customer_age')
plt.ylabel('region')
plt.title('customer_age by region')
plt.show()
```

```python
region_count = df['region'].value_counts()
plt.pie(region_count,labels=region_count.index,autopct='%1.1f%%',explode=[0.1,0,0,0,0])
plt.axis('equal')
plt.title('Records by Region')
plt.show()
```
