```python
import pandas as pd
import matplotlib.pyplot as plt
df=pd.read_csv('https://raw.githubusercontent.com/adithya-murahari/data-analysis-practice/refs/heads/main/titanic.csv')
df
```

```python
print(df.columns)
```

## here the textprops{'color':'white'} is for change the value colour inside the pie chart

```python
survived=df['survived'][df['survived']==1].count()
died=df['survived'][df['survived']==0].count()
data=[survived,died]
plt.pie(data,labels=['survived','died'],autopct='%0.1f%%',shadow=True,explode=[0,0.1],colors=['black','grey'],textprops={'color': 'white'})


plt.show()
```
